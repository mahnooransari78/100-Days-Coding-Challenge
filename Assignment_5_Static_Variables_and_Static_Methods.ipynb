```python
# 5. Static Variables and Static Methods
# Assignment:
# Create a class MathUtils with a static method add(a, b) that returns the sum. No class or instance variables should be used.

class MathUtils:
  @staticmethod
  def add(a, b):
    return a + b

# Static method ko directly class ke through call karte hain
result = MathUtils.add(12, 16)
print("Sum is: ", result)

# Output:
# Sum is:  28
```
